```python
# ========================================================================
# STEP 1: IMPORT NUMPY
# ========================================================================

# NumPy is a Python library used mainly for numerical calculations.
# We will use NumPy later for np.round(), which rounds our accuracy values.
import numpy as np

#==========================================================================
    
```

```python
# =============================================================
# STEP 2: IMPORT PANDAS
# ===========================================================

# pandas is used to read,inspect and work with tabular data.
# Think of a Pandas DataFrame as an Excel-like table inside Python.
import pandas as pd
from pathlib import Path
```

```python
#=================================================================
# STEP 3: LOAD THE DATASET
#=================================================================

# STORE THE NAME OF OUR CSV FILE IN A VARIABLE
# Keeping the file name in a variable makes the code easier to change later.
# the uploaded dataset contains 1,000 rows and 10 columns.
filepath = "advertising.csv"

# pd.read_csv() reads the CSV file and converts it into A Pandas DataFrame.
# We store the complete dataset in the variable called "df".
df = pd.read_csv("advertising.csv")
```

```python
# =====================================================================
# STEP 4 : UNDERSTAND / INSPECT THE DATA
#======================================================================

# df.info() dispalys basic information about the dataset:
# - number of rows
# - number of cloumns
# - colmns names
# - data types
# - number of non-null values
#
# We do this before ML because we should undertand our data
# before giving it to a machine Learning algorithm.
df.info()

# df.head() displays the first 5 rows of the dataset by default.
# This lets us visually check what the data looks like.
head = df.head()

# Display the first 5 rows that we stored above.
print("\nFirst 5 rows of the dataset:")
print(head)

# df.describe() gives a statistical summary of numeical columns.
# It includes values such as count, mean, standard deviation,
# minimum, maximum and quartiles.
print("\nStatistical summary of numerical columns:")
print(df.describe())
```

```python
# ==================================================================
# STEP 5: SELECT THE INPUT FEATURES (X)
# ==================================================================

# Import train_test_split from Scikit-learn.
# This function is used to divide our dataset into training and testing data.
from sklearn.model_selection import train_test_split

# X represents the INPUTS/features that the ML model will use
# to make its prediction
# We selected these five columns because they are numeric/usable
# inputs for this Logistic Regression example:
# 1.Daliy Time Spent on Site
# 2. Age
# 3.Area Income 
# 4.Daily Internet Usage
# 5. Male
#
# We are not using columns such as Ad Topic Line, City, Country 
# Timestamp in this basic example
x = df[
    [
        'Daily Time Spent on Site',
        'Age',
        'Area Income',
        'Daily Internet Usage',
        'Male'
    ]
]
        
```

```python
#===================================================================
# STEP 6: SELECT THE TARGET VARIABLE (y)
# =================================================

#y represents the OUTPUT/TARGET that we want the model to predict.

#"Clicked on Ad" contains the answer:

# 0 = The person did not click the advertisement.

# 1 = The person clicked the advertisement.

# Therefore:

#  X  = information we use to make the prediction

#  y =  answer that we want the model to predict

y = df['Clicked on Ad']


```

```python
#STEP 7: SPLIT DATA INTO TRAINING AND TESTING DATA

#train_test_split() divides our dataset into two parts:
#Training data used by the model to learn
#Testing data used to evaluate the model on unseen data
#test size-0.3 means 30% of the data is kept for testing.
#Therefore approximately:
#70% training data
#30% testing data
#
# random_state=42 makes the random split reproducible.
#The number 42 is not special; it is simply a fixed random seed.

#stratify-y keeps approximately the same proportion of target classes
#(Clicked / Not Clicked) in both training and testing data.

X_train, X_test, y_train, y_test  = train_test_split(

   x,
   y,
   test_size=0.3,
   random_state=42,
   stratify=y
)
```

```python
#===================================================================
# STEP 8: CHECK  THE SIZE / SHAPE OF THE SPLIT DATA
# ==================================================================

# shape tells us the dimensions of a Dataframe or Series.
#
# X_train and X_test have:
#   number of rows x number of features
#
# y_train and y_test have :
# number of target values
print("\nShapes after train-test split:")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
#===================================================================
# STEP 9: IMPORT LOGISTIC REGRESSION
# ==================================================================

# Logistic Regression is a machine Learning alogorithm commonly
# used for classification problems.
#
# In this project, the model predicts wheather:
#   0 = Not Clicked
#   1 = Clicked
from sklearn.linear_model import LogisticRegression
```

```python
#===================================================================
# STEP 10 : IMPORT THE ACCURACY METRIC
# ==================================================================

# accuracy_score() compares the actual answers with the model's
# predicted answers and calculates the percentage of correct predictions.
from sklearn.metrics import accuracy_score
```

```python
#===================================================================
# STEP 11 : CREATE THE LOGISTIC REGRESSION MODEL
# ==================================================================

# Here we create the Logistic Regression model.
#
# At this point the model has NOT learned from our data yet.
# It is just a model object that is ready to be trained.
#
# solver='liblinear' specifies the numerical optimization method
# used by Logistic Regression to find its model parameters.
log_reg = LogisticRegression (solver='liblinear')
```

```python
#===================================================================
# STEP 12 : TRAIN/FIT THE MODEL
# ==================================================================

# #.fit() is the training step.
#
# The model learns the relationship between:
# X_train input/features
#
#
# Simple student analogy:
# fit() "Let the student study the training questions."
log_reg.fit(X_train, y_train)
```

```python
#===================================================================
# STEP 13 : MAKE PREDICTIONS ON TRAINING DATA
# ==================================================================

#.predict() asks the trained model to make predictions.
#
# Here we give it X_train,which is the same data used for learning.
# The predictions are stored in y_train_pred.
y_train_pred = log_reg.predict(X_train)
```

```python
#===================================================================
# STEP 14 : MAKE PREDICTIONS ON TEST DATA
# ==================================================================

# Now we give the trained model X_test.
#
# X_test was Not used to train the model.
# therefore, this gives us an idea of how the model performs
# on unseen data
#
# The predictions are stored in y_test_pred.
y_test_pred = log_reg.predict(X_test)
```

```python
#===================================================================
# STEP 15 : CALCULATE TRAINING ACCURACY
# ==================================================================

# Compare:
# y-train  = actual answers from the training data
# y_train_pred = model's predictions for the training data
#
# The result tells us how accurately the model predicts
# the data is learned from
train_accuracy = accuracy_score(y_train, y_train_pred)
```

```python
#===================================================================
# STEP 16 : CALCULATE TEST ACCURACY
# ==================================================================

# Compare:
# y-test  = actual answers from unseen test data
# y_test_pred = model's predictions for the test data
#
# This is especially useful because it tells us how the model
# performs on data that it did not use during training.
test_accuracy = accuracy_score(y_test, y_test_pred)
```

```python
#===================================================================
# STEP 17 : PRINT THE  ACCURACY RESULTS
# ==================================================================

# Print the training accuracy.
# for example 0.91 means approximately 91% of training predictions
#were correct.
print(f"\nTraining Accuracy: {train_accuracy:.2f}")

# Print the test accuracy.
# for example, 0.89 means approximately 89% of test predictions
#were correct
print(f"Test Accuracy: {test_accuracy:.2f}")
```

```python
#===================================================================
# STEP 18 : SHUFFLE SPLIT CROSS-VALIDATION
# ==================================================================

# ShuffleSplit is a cross-validation method.
#
# Instead of relying on only ONE train/test split, it creates
#multiple random train/test splits.
#
# This helps us umderstand wheather the model's perfformance is 
# reasonably consistent across different random samples.
from sklearn.model_selection import ShuffleSplit, cross_validate
shuffle_split = ShuffleSplit(
    n_splits=200,
    test_size=0.3,
    random_state=42
)
```

```python
#===================================================================
# STEP 19 :  RUN SHUFFLE SPLIT CROSS-VALIDATION
# ==================================================================
cv_results = cross_validate(
    log_reg,
    x,
    y,
    cv=shuffle_split,
    scoring='accuracy',
    return_train_score=True
)
```

```python
#===================================================================
# STEP 20 :  EXTRACT SHUFFLE SPLIT TRAINING SCORES
# ==================================================================
train_scores = cv_results['train_score']
```

```python
#===================================================================
# STEP 21 :  EXTRACT SHUFFLE SPLIT TEST SCORES
# ==================================================================

# Extract all test/validation accuracy values.
#
#Again, because n_splits=200, there will be 200 test scores
test_scores = cv_results['test_score']
```

```python
#===================================================================
# STEP 22 :  CALCULATE AVERAGE SHUFFLESPLIT ACCURACY
# ==================================================================

#.mean() calculates the average of the 200 training scores.
# np.round(...,2) rounds the answer to 2 decimal places
print(
    "\nShuffleSplit - Average Training Accuracy:",
    np.round(train_scores.mean(), 2)
)
# Calculate the average of the 200 test/validation scores
# This gives us the average performance across all random splits.
print(
    "ShuffleSplit - Average Test Accuracy:",
    np.round(test_scores.mean(), 2)
)
```

```python
#===================================================================
# STEP 23 :  K-FOLD CROSS VALIDATION
# ==================================================================
# K-Fold Cross-Validation divides the dataset into K groups/folds.
#
# Here K5, so the dataset is divided into 5 parts.
#
# The model is trained and evaluated 5 times.
# In each round, a different fold is used for validation/testing,
# while the other 4 folds are used for training.
from sklearn.model_selection import KFold

#Create the K-Fold cross-validation strategy.
#
# n_splits-5:

# Divide the dataset into 5 folds.

# shuffle-True:

#  Shuffle the records before creating the folds.

# random_state-42:
# Make the shuffling reproducible.

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
#===================================================================
# STEP 24 :  RUN  K-FOLD CROSS VALIDATION
# ==================================================================

#IMPORTANT:
# The original code used:

#   cv-shuffle_split

# in this K-Fold section. That would NOT perform K-Fold.

# It would run ShuffleSplit again.

# We therefore use:

#  cv-kfold

# so that Scikit-learn actually uses the 5 K-Fold splits.

#cross_validate() returns both training and test accuracy

# for each of the 5 folds.

cv_results = cross_validate(
  log_reg,
    x,
    y,
    cv=kfold,
    scoring='accuracy',
    return_train_score=True
)

```

```python
#===================================================================
# STEP 25 :  EXTRACT K-FOLD  TRAINING SCORES
# ==================================================================

# Extract the training accuracy from each of the 5 folds
train_scores = cv_results['train_score']
```

```python
#===================================================================
# STEP 26 :  EXTRACT K-FOLD  Test SCORES
# ==================================================================

#  Extract the test/validation accuracy from each of the 5 folds
test_scores = cv_results['test_score']
```

```python
#===================================================================
# STEP 27 : CALCULATE AVERAGE K-FOLD ACCURACY
# ==================================================================

# Calculate the average training accuracy across the 5 folds.
print(
    "\nK-Fold - Average Training Accuracy:",
    np.round(train_scores.mean(), 2)
)

# Calculate the average test/validation  accuracy across the 5 folds.
print(
    "K-Fold - Average Test Accuracy:",
    np.round(test_scores.mean(), 2)
)
```
